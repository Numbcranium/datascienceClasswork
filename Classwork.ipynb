```python
import numpy as np
```

```python
# Number 1
sales = np.array([
 [20, 25, 18, 30, 22], # Rice
 [15, 10, 12, 18, 20], # Beans
 [30, 35, 28, 40, 38], # Bread
 [ 8, 12, 10, 9, 11] # Milk
])
prices = np.array([1500, 1200, 800, 600]) # ₦ per unit: Rice, Beans, Bread, Milk
sales
```

```python
# Number 1.1
np.shape(sales) # shape of sales
```

```python
# Number 1.2
sales.sum(axis = 1) # Total units sold for each product
```

```python
# Number 1.3
TotalUnit_EachDay = sales.sum(axis = 0) #Total units sold each day
TotalUnit_EachDay.argmax() # Busiest day
```

```python
# Number 1.4
Total_Quantity = sales.sum(axis = 1)
Revenue_Each = Total_Quantity * prices # Revenue for each product
Revenue_Each.sum() # Shops total revenue
```

```python
# Number 1.5
Bread_sold = sales[2]
MoreThan_Thirty = Bread_sold > 30 # Days bread sold for more than 30 units
np.sum(MoreThan_Thirty) # How many days it was
```
